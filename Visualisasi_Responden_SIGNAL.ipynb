```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Baca file CSV
df = pd.read_csv('kuesioner SIGNAL (Jawaban) (1).csv', sep=';')

# Ambil kolom pertanyaan mulai dari kolom ke-4
question_cols = df.columns[3:]

# Konversi ke numerik
df[question_cols] = df[question_cols].apply(pd.to_numeric, errors='coerce')

# Loop tiap pertanyaan
for col in question_cols:
    plt.figure(figsize=(6,4))

    # Hitung jumlah masing-masing nilai (1–4) lalu pastikan urutan
    counts = df[col].value_counts().sort_index()
    counts = counts.reindex([1,2,3,4], fill_value=0)  # pastikan semua skor muncul

    sns.barplot(x=counts.index, y=counts.values, palette='pastel')

    plt.title(f"Distribusi Jawaban: \n{col.splitlines()[0]}", wrap=True)
    plt.xlabel("Skor")
    plt.ylabel("Jumlah Responden")
    plt.xticks([0,1,2,3], ['1\n(Sangat Tidak Setuju)',
                           '2\n(Tidak Setuju)',
                           '3\n(Setuju)',
                           '4\n(Sangat Setuju)'])
    plt.grid(axis='y', linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
```
